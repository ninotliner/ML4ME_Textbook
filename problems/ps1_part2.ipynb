### Task 1
Assuming that the 7-dimensional space is excessive, you would like to reduce the dimension of the space. However, what dimensionality of space should we reduce it to? To determine this we need to compute its intrinsic dimensionality. Plot the relative value of the information content of each of the principal components and compare them.

Note: this information content is called the “explained variance” of each component, but you can also get this from the magnitude of the singular values. This plot is sometimes called a “Scree Plot”.

```python
# Code Here
# compute svd
import numpy as np

U, s, Vt = np.linalg.svd(data, full_matrices=False)
print("U shape:", U.shape, "s shape:", s.shape, "Vt shape:", Vt.shape)
print("Singular values:", s)
# s = np.sort(s)[::-1]

# plot the singular values against their index
import matplotlib.pyplot as plt
plt.plot(s, 'o')
plt.xlabel('Index')
plt.ylabel('Singular Value')
```

**Question:** Approximately how many components dominate the space?, and what does this tell us about the intrinsic dimensionality of the space?

**Response**: Two components dominate the transformation, while the remaining components are almost negligible. This means that any input vector is primarily influenced by its components along two particular input directions, which are strongly scaled and mapped to two corresponding output directions. Components along the remaining input directions are strongly suppressed. As a result, only two directions are significant in the input space, and the output of the transformation can be approximated as a combination of two particular orthogonal output directions.

#### Task 2
Now use PCA to project the 7-dimensional points on the K-dimensional space (where K is your answer from above) and plot the points. (For K=1,2, or 3, use a 1, 2, or 3D plot, respectively. For 4+ dimensions, use a grid of pairwise 2D Plots, like the Scatter Matrix we used in class).

```python
# Code Here

for K in range (1, 7):
    # reconstruct the data using the first K singular values
    S = np.zeros((K, K))
    np.fill_diagonal(S, s[:K])
    X_reconstructed = U[:, :K] @ S @ Vt[:K, :]
    
    # Coordinates in the retained principal-component space.
    Z = U[:, :K] * s[:K]

    # Plot the reconstructed data in its K-dimensional PCA space.
    if K in [1, 2, 3]:
        if K == 1:
            fig, ax = plt.subplots(figsize=(8, 2.5))
            ax.plot(Z[:, 0], np.zeros(Z.shape[0]), '.', alpha=0.35)
            ax.set(yticks=[], xlabel='PC 1', title='Rank-1 reconstruction')
        elif K == 2:
            fig, ax = plt.subplots(figsize=(6, 5))
            ax.scatter(Z[:, 0], Z[:, 1], s=8, alpha=0.35)
            ax.set(xlabel='PC 1', ylabel='PC 2', title='Rank-2 reconstruction')
            ax.axis('equal')
        else:
            fig = plt.figure(figsize=(7, 5))
            ax = fig.add_subplot(111, projection='3d')
            ax.scatter(Z[:, 0], Z[:, 1], Z[:, 2], s=8, alpha=0.35)
            ax.set(xlabel='PC 1', ylabel='PC 2', zlabel='PC 3',
                   title='Rank-3 reconstruction')
        plt.tight_layout()
        plt.show()
    else:
        axes = pandas.plotting.scatter_matrix(
            pandas.DataFrame(Z, columns=[f'PC {i + 1}' for i in range(K)]),
            figsize=(10, 10), diagonal='none', alpha=0.35, s=8)
        plt.suptitle(f'Rank-{K} reconstruction')
        plt.show()
```

**Question:** What do you notice?

**Response**: We see that principle components 1 and 2 form an image, which was probably what the data was intended for. There is no corrolation when we look at the other components, suggesting they are just noise.
